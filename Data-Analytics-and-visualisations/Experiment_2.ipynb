```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "order data  ": ["2024-01-10", "2024-01-15", "2024-02-05", "2024-02-12", "2024-03-01", "2024-03-15", "2024-04-10", "2024-04-20", "2024-05-05", "2024-01-15"],
    "Product name": ["Laptop", "Mouse", "Printer", "Tablet", "Laptop", "Monitor", "Mouse", "Printer", "Smartwatch", "mouse"],
    "Category": ["Electronics", "Accessories", "office", "Electronics", "Electronics", "Accessories", "Accessories", "office", "Electronics", "Accessories"],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West", "North", "South"],
    "Quantity": [2, 5, 3, 4, 1, 2, 6, 2, 3, 5],
    "Sales": [50000, 2500, 9000, 30000, 55000, 16000, 3000, 6000, 45000, 2500],
    "profit": [10000, 500, 1500, 6000, 11000, 3000, None, 1000, 9000, 500]
})
```

```python
print("Original Dataset:")
print(df)
```

```python
print("First 5 Rows:")
print(df.head())
```

```python
print("Last 5 Rows:")
print(df.tail())
```

```python
print("Shape of Dataset:")
print(df.shape)

print("Column Names:")
print(df.columns)

print("Data Types:")
print(df.dtypes)
```

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("Rows with Missing Values:")
print(df[df.isnull().any(axis=1)])
```

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())

print("Duplicate Rows:")
print(df[df.duplicated()])
```

```python
new_sales = pd.DataFrame({
    "order data  ": ["2024-06-01", "2024-06-05", "2024-06-10"],
    "Product name": ["Laptop", "Mouse", "Printer"],
    "Category": ["Electronics", "Accessories", "office"],
    "Region": ["North", "South", "East"],
    "Quantity": [2, 5, 3],
    "Sales": [60000, 2000, 9000],
    "profit": [12000, 400, 1800]
})

print("New Sales Dataset:")
print(new_sales)
```

```python
combined_df = pd.concat([df, new_sales], ignore_index=True)

print("Concatenated Dataset:")
print(combined_df)
```

```python
product_info = pd.DataFrame({
    "Product name": ["Laptop", "Mouse", "Printer", "Tablet", "Monitor"],
    "Supplier": ["Dell", "Logitech", "HP", "Samsung", "LG"],
    "Discount": [10, 5, 15, 8, 12]
})

print("Product Information:")
print(product_info)
```

```python
merge_df = pd.merge(df, product_info, on="Product name", how="left")

print("Merged Dataset:")
print(merge_df)
```

```python
region_info = pd.DataFrame({
    " Region_Manager": ["Manager A", "Manager B", "Manager C", "Manager D"]
}, index=["North", "South", "East", "West"])

print("Region Information:")
print(region_info)
```

```python
df_region = df.set_index("Region")
joined_df = df_region.join(region_info)

print("Joined Dataset:")
print(joined_df)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print("Total Sales by Region:")
print(region_sales)
```

```python
pivot_table = pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum",
    fill_value=0
)

print("Pivot Table:")
print(pivot_table)
```
